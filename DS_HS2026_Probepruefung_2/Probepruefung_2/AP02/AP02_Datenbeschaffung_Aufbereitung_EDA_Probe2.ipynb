````python
# Libraries
import os
import re
import numpy as np
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
from bs4 import BeautifulSoup

# Ignore warnings
import warnings
warnings.filterwarnings('ignore')

# Get current working directory
print(os.getcwd())
````

### Ergebnisseiten scrapen → `df_raw`

Der folgende Code liest die drei gespeicherten Ergebnisseiten mit BeautifulSoup ein (analog zur Übung `skipass_data.ipynb`). Jede Unterkunft (`<div class="listing-card">`) wird zu einer Zeile im Data Frame `df_raw`. Die Werte werden **als Text** übernommen (Spalten mit Endung `_raw`) und noch nicht bereinigt.

````python
# Initialize list for the extracted rows
rows = []

# Loop over the three result pages
for page in [1, 2, 3]:

    # Read html from file
    with open(f'html/listing_page_{page}.html', 'r', encoding='utf-8') as file:
        html_content = file.read()

    # Parse the HTML content
    soup = BeautifulSoup(html_content, 'html.parser')

    # Locate the listing cards and extract values
    for card in soup.select('div.listing-card'):
        row_data = [card['data-listing-id'],
                    card.select('span.type')[0].get_text().replace('\xa0', ' ').strip(),
                    card.select('span.location')[0].get_text().replace('\xa0', ' ').strip(),
                    card.select('li.rooms')[0].get_text().replace('\xa0', ' ').strip(),
                    card.select('li.area')[0].get_text().replace('\xa0', ' ').strip(),
                    card.select('li.guests')[0].get_text().replace('\xa0', ' ').strip(),
                    card.select('li.lift')[0].get_text().replace('\xa0', ' ').strip(),
                    card.select('span.host-type')[0].get_text().replace('\xa0', ' ').strip(),
                    card.select('span.price')[0].get_text().replace('\xa0', ' ').strip()]
        rows.append(row_data)

# Create a DataFrame
df_raw = pd.DataFrame(rows, columns=['listing_id', 'type', 'location', 'rooms_raw', 'area_raw',
                                     'guests_raw', 'lift_raw', 'host_type', 'price_raw'])

# Show dimensions and first rows
print('Number of rows and columns:', df_raw.shape)
df_raw.head()
````
