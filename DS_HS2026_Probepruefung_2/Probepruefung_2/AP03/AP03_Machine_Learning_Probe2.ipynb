## Setup (wird nicht bewertet)

````python
# Libraries
import os
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import statsmodels.api as sm
from sklearn import tree
from sklearn.model_selection import train_test_split
from sklearn.tree import DecisionTreeRegressor, DecisionTreeClassifier
from sklearn.ensemble import RandomForestRegressor, RandomForestClassifier
from sklearn.metrics import r2_score, accuracy_score, confusion_matrix, RocCurveDisplay

# Ignore warnings
import warnings
warnings.filterwarnings('ignore')

# Get current working directory
print(os.getcwd())
````

### Kundendaten laden → `df`

````python
# Import data from csv to pandas dataframe named 'df'
df = pd.read_csv('telecom_data.csv')

# Show dimensions and first rows
print('Number of rows and columns:', df.shape)
df.head()
````

### Trainings- und Testdaten erstellen

Der folgende Code teilt die Daten **zweimal** in 80 % Trainingsdaten und 20 % Testdaten auf:

- **Rechnung (Regression):** `X_train`, `X_test`, `y_train`, `y_test` mit den Features in `bill_features` und der Zielvariable `monthly_bill_chf`.
- **Kündigung (Klassifikation):** `X_train_class`, `X_test_class`, `y_train_class`, `y_test_class` mit den Features in `churn_features` und der Zielvariable `churn`. `stratify` sorgt dafür, dass der Anteil der Kündiger in Trainings- und Testdaten gleich gross ist.

````python
# Features and target for the bill prediction (regression)
bill_features = ['data_gb', 'call_minutes', 'has_tv', 'lines']
X_train, X_test, y_train, y_test = train_test_split(df[bill_features],
                                                    df['monthly_bill_chf'],
                                                    test_size=0.20,
                                                    random_state=42)

# Features and target for the churn prediction (classification)
churn_features = ['support_calls', 'contract_months_left', 'price_increase_pct']
X_train_class, X_test_class, y_train_class, y_test_class = train_test_split(df[churn_features],
                                                                            df['churn'],
                                                                            test_size=0.20,
                                                                            random_state=42,
                                                                            stratify=df['churn'])

# Show the size of the samples
print('Regression:     ', X_train.shape[0], 'Trainingsdaten /', X_test.shape[0], 'Testdaten')
print('Klassifikation: ', X_train_class.shape[0], 'Trainingsdaten /', X_test_class.shape[0], 'Testdaten')
````
